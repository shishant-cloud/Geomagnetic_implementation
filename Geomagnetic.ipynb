```python
def show_tree(path, prefix=""):
    items = sorted(os.listdir(path))

    for i, item in enumerate(items):
        full_path = os.path.join(path, item)
        connector = "└── " if i == len(items) - 1 else "├── "

        print(prefix + connector + item)

        if os.path.isdir(full_path):
            extension = "    " if i == len(items) - 1 else "│   "
            show_tree(full_path, prefix + extension)


print("Geomagnetic_Project/")
show_tree(PROJECT_DIR)
```

Build a parser for one .min file

```python
import pandas as pd
import re

def read_iaga_min(file_path):
    """
    Read a KAK IAGA-2002 .min file.
    """

    rows = []

    with open(file_path, "r", errors="ignore") as f:
        for line in f:

            # Skip header and comments
            if line.startswith("#"):
                continue

            # Skip the column-name line
            if line.startswith("DATE"):
                continue

            parts = line.split()

            # Expected:
            # DATE TIME DOY KAKX KAKY KAKZ KAKG
            if len(parts) < 7:
                continue

            try:
                date = parts[0]
                time = parts[1]

                rows.append({
                    "datetime": pd.to_datetime(
                        date + " " + time
                    ),
                    "KAKX": float(parts[3]),
                    "KAKY": float(parts[4]),
                    "KAKZ": float(parts[5]),
                    "KAKG": float(parts[6])
                })

            except (ValueError, TypeError):
                continue

    return pd.DataFrame(rows)
```

# Create a reusable parser function

```python
import pandas as pd
import os


def parse_kak_min_file(file_path):
    """
    Parse one KAK IAGA-2002 .min file.

    Parameters
    ----------
    file_path : str
        Path to the .min file.

    Returns
    -------
    pandas.DataFrame
        Clean dataframe containing:
        datetime, KAKX, KAKY, KAKZ, KAKG
    """

    # ---------------------------------------------------------
    # 1. Read the file and locate the DATE header
    # ---------------------------------------------------------
    with open(file_path, "r", errors="ignore") as f:
        lines = f.readlines()

    header_line = None

    for i, line in enumerate(lines):
        if line.strip().startswith("DATE"):
            header_line = i
            break

    if header_line is None:
        raise ValueError(
            f"DATE header not found: {file_path}"
        )

    # ---------------------------------------------------------
    # 2. Read measurement data
    # ---------------------------------------------------------
    df = pd.read_csv(
        file_path,
        sep=r"\s+",
        skiprows=header_line + 1,
        names=[
            "DATE",
            "TIME",
            "DOY",
            "KAKX",
            "KAKY",
            "KAKZ",
            "KAKG"
        ],
        engine="python"
    )

    # ---------------------------------------------------------
    # 3. Create timestamp
    # ---------------------------------------------------------
    df["datetime"] = pd.to_datetime(
        df["DATE"] + " " + df["TIME"],
        errors="coerce"
    )

    # ---------------------------------------------------------
    # 4. Convert measurements to numeric
    # ---------------------------------------------------------
    measurement_columns = [
        "KAKX",
        "KAKY",
        "KAKZ",
        "KAKG"
    ]

    for column in measurement_columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

    # ---------------------------------------------------------
    # 5. Keep only required columns
    # ---------------------------------------------------------
    df = df[
        [
            "datetime",
            "KAKX",
            "KAKY",
            "KAKZ",
            "KAKG"
        ]
    ]

    # ---------------------------------------------------------
    # 6. Remove completely invalid timestamps
    # ---------------------------------------------------------
    df = df.dropna(
        subset=["datetime"]
    )

    # ---------------------------------------------------------
    # 7. Sort chronologically
    # ---------------------------------------------------------
    df = df.sort_values(
        "datetime"
    ).reset_index(drop=True)

    return df
```

# Add a quality-control function

Now we'll create a separate function for QC.

This is important because parsing and quality control are two different operations.

```python
def quality_check_kak(df):
    """
    Perform basic quality checks on a parsed KAK dataframe.
    """

    report = {}

    # Number of records
    report["rows"] = len(df)

    # Missing values
    report["missing_datetime"] = int(
        df["datetime"].isna().sum()
    )

    report["missing_KAKX"] = int(
        df["KAKX"].isna().sum()
    )

    report["missing_KAKY"] = int(
        df["KAKY"].isna().sum()
    )

    report["missing_KAKZ"] = int(
        df["KAKZ"].isna().sum()
    )

    report["missing_KAKG"] = int(
        df["KAKG"].isna().sum()
    )

    # Duplicate timestamps
    report["duplicate_timestamps"] = int(
        df["datetime"].duplicated().sum()
    )

    # Start/end
    report["start"] = df["datetime"].min()
    report["end"] = df["datetime"].max()

    # Time gaps
    if len(df) > 1:

        differences = df["datetime"].diff().dropna()

        report["one_minute_intervals"] = int(
            (differences == pd.Timedelta(minutes=1)).sum()
        )

        report["non_one_minute_intervals"] = int(
            (differences != pd.Timedelta(minutes=1)).sum()
        )

    else:

        report["one_minute_intervals"] = 0
        report["non_one_minute_intervals"] = 0

    return report
```

```python
def parse_kak_min_file(file_path):

    with open(
        file_path,
        "r",
        errors="ignore"
    ) as f:

        lines = f.readlines()

    # Find the measurement header
    header_index = None

    for i, line in enumerate(lines):

        if line.strip().startswith(
            "DATE       TIME"
        ):
            header_index = i
            break

    if header_index is None:
        raise ValueError(
            f"DATE header not found: {file_path}"
        )

    data = []

    for line in lines[header_index + 1:]:

        parts = line.split()

        if len(parts) < 7:
            continue

        try:

            date = parts[0]
            time = parts[1]

            x = float(parts[3])
            y = float(parts[4])
            z = float(parts[5])
            g = float(parts[6])

            data.append([
                date,
                time,
                x,
                y,
                z,
                g
            ])

        except (ValueError, IndexError):
            continue

    df = pd.DataFrame(
        data,
        columns=[
            "date",
            "time",
            "KAKX",
            "KAKY",
            "KAKZ",
            "KAKG"
        ]
    )

    df["datetime"] = pd.to_datetime(
        df["date"] + " " + df["time"],
        errors="coerce"
    )

    df = df.dropna(
        subset=["datetime"]
    )

    df = df[
        [
            "datetime",
            "KAKX",
            "KAKY",
            "KAKZ",
            "KAKG"
        ]
    ]

    df = df.sort_values(
        "datetime"
    )

    return df.reset_index(drop=True)
```
